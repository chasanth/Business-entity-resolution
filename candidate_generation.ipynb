```python
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.neighbors import NearestNeighbors


def normalize_name(x):
    if pd.isna(x):
        return ""

    x = str(x).lower()

    replacements = {
        "private limited": "pvt ltd",
        "private ltd": "pvt ltd",
        "corporation": "corp",
        "incorporated": "inc",
        "limited": "ltd",
        "company": "co"
    }

    for old, new in replacements.items():
        x = x.replace(old, new)

    x = re.sub(r"[^a-z0-9\s]", " ", x)
    x = re.sub(r"\s+", " ", x).strip()

    return x


s1["name_norm"] = s1["business_name"].map(normalize_name)
s2["name_norm"] = s2["business_name"].map(normalize_name)
s3["name_norm"] = s3["business_name"].map(normalize_name)


# Combine S2 and S3 for one candidate index
candidate_pool = pd.concat(
    [
        s2[["entity_id", "name_norm", "country"]].assign(source="S2"),
        s3[["entity_id", "name_norm", "country"]].assign(source="S3")
    ],
    ignore_index=True
)

print("Candidate pool:", len(candidate_pool))
```

```python
vectorizer = TfidfVectorizer(
    analyzer="char",
    ngram_range=(2, 5),
    min_df=2,
    max_features=200000,
    sublinear_tf=True
)

candidate_matrix = vectorizer.fit_transform(
    candidate_pool["name_norm"]
)

s1_matrix = vectorizer.transform(
    s1["name_norm"]
)

print("S1 matrix:", s1_matrix.shape)
print("Candidate matrix:", candidate_matrix.shape)
```

```python

TOP_K = 10

nn = NearestNeighbors(
    n_neighbors=TOP_K,
    metric="cosine",
    algorithm="brute",
    n_jobs=-1
)

nn.fit(candidate_matrix)

distances, indices = nn.kneighbors(s1_matrix)

print("Candidate search complete.")
print("Distance shape:", distances.shape)
print("Index shape:", indices.shape)
```

```python
# Separate name and address candidate generation

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.neighbors import NearestNeighbors

# ---------------------------------
# 1. Normalize name and address
# ---------------------------------

s1["name_norm"] = s1["business_name"].map(normalize_text)
s2["name_norm"] = s2["business_name"].map(normalize_text)
s3["name_norm"] = s3["business_name"].map(normalize_text)

s1["address_norm"] = s1["business_address"].map(normalize_text)
s2["address_norm"] = s2["business_address"].map(normalize_text)
s3["address_norm"] = s3["business_address"].map(normalize_text)


# ---------------------------------
# 2. NAME TF-IDF
# ---------------------------------

name_vectorizer = TfidfVectorizer(
    analyzer="char",
    ngram_range=(2, 5),
    min_df=2,
    max_features=200000,
    sublinear_tf=True
)

name_pool = pd.concat(
    [
        s2[["entity_id", "name_norm", "country"]].assign(source="S2"),
        s3[["entity_id", "name_norm", "country"]].assign(source="S3")
    ],
    ignore_index=True
)

name_matrix = name_vectorizer.fit_transform(
    name_pool["name_norm"]
)

s1_name_matrix = name_vectorizer.transform(
    s1["name_norm"]
)

name_nn = NearestNeighbors(
    n_neighbors=10,
    metric="cosine",
    algorithm="brute",
    n_jobs=-1
)

name_nn.fit(name_matrix)

name_distances, name_indices = name_nn.kneighbors(
    s1_name_matrix
)

print("Name candidates generated.")
```

```python
# ---------------------------------
# 3. ADDRESS TF-IDF
# ---------------------------------

address_vectorizer = TfidfVectorizer(
    analyzer="char",
    ngram_range=(3, 5),
    min_df=2,
    max_features=200000,
    sublinear_tf=True
)

address_pool = name_pool.copy()

address_pool["address_norm"] = pd.concat(
    [
        s2["address_norm"],
        s3["address_norm"]
    ],
    ignore_index=True
)

address_matrix = address_vectorizer.fit_transform(
    address_pool["address_norm"]
)

s1_address_matrix = address_vectorizer.transform(
    s1["address_norm"]
)

address_nn = NearestNeighbors(
    n_neighbors=10,
    metric="cosine",
    algorithm="brute",
    n_jobs=-1
)

address_nn.fit(address_matrix)

address_distances, address_indices = address_nn.kneighbors(
    s1_address_matrix
)

print("Address candidates generated.")
```

```python
# ---------------------------------
# 4. UNION NAME + ADDRESS CANDIDATES
# ---------------------------------

candidate_rows = []

for i in range(len(s1)):

    s1_id = s1.iloc[i]["entity_id"]

    # Name candidates
    for rank in range(10):

        idx = name_indices[i, rank]

        candidate_rows.append({
            "source1_entity_id": s1_id,
            "candidate_entity_id": name_pool.iloc[idx]["entity_id"],
            "candidate_source": name_pool.iloc[idx]["source"],
            "channel": "name",
            "similarity": 1 - name_distances[i, rank]
        })

    # Address candidates
    for rank in range(10):

        idx = address_indices[i, rank]

        candidate_rows.append({
            "source1_entity_id": s1_id,
            "candidate_entity_id": address_pool.iloc[idx]["entity_id"],
            "candidate_source": address_pool.iloc[idx]["source"],
            "channel": "address",
            "similarity": 1 - address_distances[i, rank]
        })


candidates = pd.DataFrame(candidate_rows)

# Remove duplicate S1-candidate pairs
candidates = (
    candidates
    .sort_values("similarity", ascending=False)
    .drop_duplicates(
        ["source1_entity_id", "candidate_entity_id"]
    )
)

print("Total candidate pairs:", len(candidates))
print(candidates.head(20))
```

```python
# Calculate blocking recall

# Create lookup:
# S1 ID -> set of true matched S2/S3 IDs
ground_truth = {}

for _, row in gt.iterrows():
    s1_id = row["source1_entity_id"]
    
    if pd.isna(row["matched_entity_ids"]):
        ground_truth[s1_id] = set()
    else:
        ground_truth[s1_id] = set(
            str(row["matched_entity_ids"]).split(",")
        )

# Candidate lookup
candidate_lookup = (
    candidates
    .groupby("source1_entity_id")["candidate_entity_id"]
    .apply(set)
    .to_dict()
)

# Calculate recall
total_true = 0
found_true = 0

for s1_id, true_ids in ground_truth.items():
    
    # Only evaluate S1s present in our current sample
    if s1_id not in candidate_lookup:
        continue
    
    predicted_candidates = candidate_lookup[s1_id]
    
    total_true += len(true_ids)
    found_true += len(true_ids & predicted_candidates)

recall = found_true / total_true if total_true > 0 else 0

print("True matches evaluated:", total_true)
print("True matches found:", found_true)
print("Blocking Recall:", round(recall, 4))
print("Blocking Recall %:", round(recall * 100, 2), "%")
```

```python
import pandas as pd
import re
from rapidfuzz import process, fuzz

# Full training file paths
TRAIN_DIR = Path.home() / "Downloads/student_resource 2/dataset/train"

# First 100 S1 records
recall_s1 = train_s1.head(100).copy()

recall_s1["name_norm"] = recall_s1["business_name"].map(normalize_text)
recall_s1["address_norm"] = recall_s1["business_address"].map(normalize_text)

# Store best candidates found across ALL chunks
best_candidates = {
    row["entity_id"]: {}
    for _, row in recall_s1.iterrows()
}

# Process S2 and S3 separately
source_files = [
    ("S2", TRAIN_DIR / "train_source2.tsv"),
    ("S3", TRAIN_DIR / "train_source3.tsv")
]

CHUNK_SIZE = 100_000
TOP_K = 20

for source_name, file_path in source_files:

    print(f"\nProcessing {source_name}...")

    for chunk_no, chunk in enumerate(
        pd.read_csv(
            file_path,
            sep="\t",
            dtype=str,
            chunksize=CHUNK_SIZE
        )
    ):

        chunk["name_norm"] = chunk["business_name"].map(normalize_text)
        chunk["address_norm"] = chunk["business_address"].map(normalize_text)

        # Remove empty names/addresses
        name_choices = dict(
            zip(
                chunk["entity_id"],
                chunk["name_norm"]
            )
        )

        address_choices = dict(
            zip(
                chunk["entity_id"],
                chunk["address_norm"]
            )
        )

        for _, s1row in recall_s1.iterrows():

            s1_id = s1row["entity_id"]

            # -------------------------
            # NAME BLOCKING
            # -------------------------

            if s1row["name_norm"]:

                matches = process.extract(
                    s1row["name_norm"],
                    name_choices,
                    scorer=fuzz.ratio,
                    limit=TOP_K
                )

                for _, score, entity_id in matches:

                    old = best_candidates[s1_id].get(entity_id, 0)

                    if score > old:
                        best_candidates[s1_id][entity_id] = score


            # -------------------------
            # ADDRESS BLOCKING
            # -------------------------

            if s1row["address_norm"]:

                matches = process.extract(
                    s1row["address_norm"],
                    address_choices,
                    scorer=fuzz.ratio,
                    limit=TOP_K
                )

                for _, score, entity_id in matches:

                    old = best_candidates[s1_id].get(entity_id, 0)

                    if score > old:
                        best_candidates[s1_id][entity_id] = score

        if (chunk_no + 1) % 10 == 0:
            print(
                f"  processed {(chunk_no + 1) * CHUNK_SIZE:,} rows"
            )

    print(f"{source_name} complete.")
```

```python
def normalize_text_unicode(x):
    if pd.isna(x):
        return ""

    x = str(x).lower()

    replacements = {
        "private limited": "pvt ltd",
        "private ltd": "pvt ltd",
        "corporation": "corp",
        "incorporated": "inc",
        "limited": "ltd",
        "company": "co",
        "street": "st",
        "road": "rd",
        "avenue": "ave",
        "boulevard": "blvd",
        "drive": "dr",
        "lane": "ln",
        "highway": "hwy",
        "parkway": "pkwy",
        "apartment": "apt",
        "suite": "ste"
    }

    for old, new in replacements.items():
        x = x.replace(old, new)

    # Keep Unicode letters/numbers
    x = re.sub(r"[^\w\s]", " ", x, flags=re.UNICODE)
    x = re.sub(r"\s+", " ", x).strip()

    return x
```

```python
import re

def extract_numbers(text):
    if pd.isna(text):
        return set()

    text = str(text)

    # Extract individual numeric components.
    # Example:
    # 1056-1060 -> 1056, 1060
    # 20-6-3/12 -> 20, 6, 3, 12
    # 9/2113 -> 9, 2113
    numbers = re.findall(r'\d+', text)

    return set(numbers)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

TOP_K_NAME = 20
TOP_K_ADDRESS = 20

# Full S2 + S3 pool
train_s23 = pd.concat([train_s2, train_s3], ignore_index=True)

print("S2 + S3 rows:", len(train_s23))

# Name TF-IDF
name_vectorizer_eval = TfidfVectorizer(
    analyzer="char",
    ngram_range=(2, 5),
    min_df=2
)

name_matrix_eval = name_vectorizer_eval.fit_transform(
    train_s23["name_norm"].fillna("")
)

s1_name_eval = name_vectorizer_eval.transform(
    s1_eval["name_norm"].fillna("")
)

# Address TF-IDF
address_vectorizer_eval = TfidfVectorizer(
    analyzer="char",
    ngram_range=(2, 5),
    min_df=2
)

address_matrix_eval = address_vectorizer_eval.fit_transform(
    train_s23["address_norm"].fillna("")
)

s1_address_eval = address_vectorizer_eval.transform(
    s1_eval["address_norm"].fillna("")
)

# Candidate generation
candidate_sets_eval = []

for i in range(len(s1_eval)):

    name_scores = cosine_similarity(
        s1_name_eval[i],
        name_matrix_eval
    ).ravel()

    name_idx = np.argpartition(
        name_scores, -TOP_K_NAME
    )[-TOP_K_NAME:]

    address_scores = cosine_similarity(
        s1_address_eval[i],
        address_matrix_eval
    ).ravel()

    address_idx = np.argpartition(
        address_scores, -TOP_K_ADDRESS
    )[-TOP_K_ADDRESS:]

    indices = set(name_idx) | set(address_idx)

    # Country filter
    country = str(s1_eval.iloc[i]["country"])

    indices = [
        j for j in indices
        if str(train_s23.iloc[j]["country"]) == country
    ]

    candidate_sets_eval.append(
        set(train_s23.iloc[j]["entity_id"] for j in indices)
    )

print("\nCandidate generation completed.")
print(
    "Average candidates/S1:",
    round(np.mean([len(x) for x in candidate_sets_eval]), 2)
)
print(
    "Maximum candidates:",
    max(len(x) for x in candidate_sets_eval)
)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import normalize
import numpy as np
import pandas as pd

TOP_K = 20
CHUNK_SIZE = 100_000

# Evaluation S1
eval_names = s1_eval["name_norm"].fillna("").tolist()
eval_addresses = s1_eval["address_norm"].fillna("").tolist()

# Fit vectorizers ONLY on the 100 S1 evaluation records.
# This keeps the vocabulary small.
name_vectorizer = TfidfVectorizer(
    analyzer="char",
    ngram_range=(2, 5),
    min_df=1
)

address_vectorizer = TfidfVectorizer(
    analyzer="char",
    ngram_range=(2, 5),
    min_df=1
)

s1_name_matrix = name_vectorizer.fit_transform(eval_names)
s1_address_matrix = address_vectorizer.fit_transform(eval_addresses)

# Store best candidates for each S1
name_best = [[] for _ in range(len(s1_eval))]
address_best = [[] for _ in range(len(s1_eval))]

# --------------------------------------------------
# Process S2 and S3 separately in chunks
# --------------------------------------------------

sources = [
    ("S2", train_s2),
    ("S3", train_s3)
]

for source_name, source_df in sources:

    print(f"\nProcessing {source_name}: {len(source_df):,} rows")

    for start in range(0, len(source_df), CHUNK_SIZE):

        end = min(start + CHUNK_SIZE, len(source_df))
        chunk = source_df.iloc[start:end]

        # ---------- NAME ----------
        chunk_name_matrix = name_vectorizer.transform(
            chunk["name_norm"].fillna("")
        )

        name_scores = s1_name_matrix @ chunk_name_matrix.T

        for i in range(len(s1_eval)):

            scores = name_scores[i].toarray().ravel()

            k = min(TOP_K, len(scores))
            idx = np.argpartition(scores, -k)[-k:]

            for j in idx:
                if scores[j] > 0:
                    name_best[i].append(
                        (
                            float(scores[j]),
                            chunk.iloc[j]["entity_id"],
                            str(chunk.iloc[j]["country"])
                        )
                    )

        # ---------- ADDRESS ----------
        chunk_address_matrix = address_vectorizer.transform(
            chunk["address_norm"].fillna("")
        )

        address_scores = s1_address_matrix @ chunk_address_matrix.T

        for i in range(len(s1_eval)):

            scores = address_scores[i].toarray().ravel()

            k = min(TOP_K, len(scores))
            idx = np.argpartition(scores, -k)[-k:]

            for j in idx:
                if scores[j] > 0:
                    address_best[i].append(
                        (
                            float(scores[j]),
                            chunk.iloc[j]["entity_id"],
                            str(chunk.iloc[j]["country"])
                        )
                    )

        if end % 500_000 == 0 or end == len(source_df):
            print(f"  Processed {end:,}/{len(source_df):,}")

# --------------------------------------------------
# Keep only global top-20 per channel + country filter
# --------------------------------------------------

candidate_sets_eval = []

for i in range(len(s1_eval)):

    country = str(s1_eval.iloc[i]["country"])

    name_candidates = sorted(
        [
            x for x in name_best[i]
            if x[2] == country
        ],
        reverse=True
    )[:TOP_K]

    address_candidates = sorted(
        [
            x for x in address_best[i]
            if x[2] == country
        ],
        reverse=True
    )[:TOP_K]

    candidates = set(
        x[1] for x in name_candidates
    ) | set(
        x[1] for x in address_candidates
    )

    candidate_sets_eval.append(candidates)

print("\n========================================")
print("CANDIDATE GENERATION COMPLETED")
print("========================================")
print(
    "Average candidates/S1:",
    round(np.mean([len(x) for x in candidate_sets_eval]), 2)
)
print(
    "Maximum candidates:",
    max(len(x) for x in candidate_sets_eval)
)
```

```python
import re

def extract_numbers(text):
    if pd.isna(text):
        return set()
    return set(re.findall(r"\d+", str(text)))

# Add numbers to the 22 missed cases
diagnostic["s1_numbers"] = diagnostic["s1_address"].apply(
    extract_numbers
)

diagnostic["true_numbers"] = diagnostic["address_norm"].apply(
    extract_numbers
)

diagnostic["shared_numbers"] = diagnostic.apply(
    lambda r: r["s1_numbers"] & r["true_numbers"],
    axis=1
)

print("Missed pairs with shared address numbers:")
print(
    diagnostic[
        ["s1_id", "true_id", "shared_numbers"]
    ]
)

print(
    "\nRecovered by shared number:",
    (diagnostic["shared_numbers"].apply(len) > 0).sum()
)
```

```python
# ============================================================
# MEMORY-SAFE NAME + ADDRESS CANDIDATE GENERATION
# TOP-K = 50
# ============================================================

import numpy as np
import pandas as pd

from sklearn.feature_extraction.text import TfidfVectorizer


# ------------------------------------------------------------
# SETTINGS
# ------------------------------------------------------------

TOP_K = 50
CHUNK_SIZE = 100_000

print("TOP_K:", TOP_K)
print("CHUNK_SIZE:", CHUNK_SIZE)


# ------------------------------------------------------------
# 1. EVALUATION S1
# ------------------------------------------------------------

s1_eval = train_s1.head(100).copy().reset_index(drop=True)

print("\nEvaluation S1:", s1_eval.shape)


# ------------------------------------------------------------
# 2. PREPARE TEXT
# ------------------------------------------------------------

eval_names = (
    s1_eval["name_norm"]
    .fillna("")
    .astype(str)
    .tolist()
)

eval_addresses = (
    s1_eval["address_norm"]
    .fillna("")
    .astype(str)
    .tolist()
)


# ------------------------------------------------------------
# 3. FIT SMALL VECTORIZERS
# ------------------------------------------------------------

print("\nFitting name vectorizer...")

name_vectorizer = TfidfVectorizer(
    analyzer="char",
    ngram_range=(2, 5),
    min_df=1
)

s1_name_matrix = name_vectorizer.fit_transform(
    eval_names
)

print(
    "Name matrix:",
    s1_name_matrix.shape
)


print("\nFitting address vectorizer...")

address_vectorizer = TfidfVectorizer(
    analyzer="char",
    ngram_range=(2, 5),
    min_df=1
)

s1_address_matrix = address_vectorizer.fit_transform(
    eval_addresses
)

print(
    "Address matrix:",
    s1_address_matrix.shape
)


# ------------------------------------------------------------
# 4. STORAGE FOR BEST CANDIDATES
# ------------------------------------------------------------

name_best = [[] for _ in range(len(s1_eval))]
address_best = [[] for _ in range(len(s1_eval))]


# ------------------------------------------------------------
# 5. PROCESS S2 + S3 IN CHUNKS
# ------------------------------------------------------------

sources = [
    ("S2", train_s2),
    ("S3", train_s3)
]


for source_name, source_df in sources:

    print("\n" + "=" * 60)
    print(f"Processing {source_name}: {len(source_df):,} rows")
    print("=" * 60)

    for start in range(
        0,
        len(source_df),
        CHUNK_SIZE
    ):

        end = min(
            start + CHUNK_SIZE,
            len(source_df)
        )

        chunk = source_df.iloc[start:end]

        # ----------------------------------------------------
        # NAME
        # ----------------------------------------------------

        chunk_name_matrix = name_vectorizer.transform(
            chunk["name_norm"]
            .fillna("")
            .astype(str)
        )

        name_scores = (
            s1_name_matrix @
            chunk_name_matrix.T
        )

        for i in range(len(s1_eval)):

            scores = name_scores[i].toarray().ravel()

            k = min(
                TOP_K,
                len(scores)
            )

            if k == 0:
                continue

            idx = np.argpartition(
                scores,
                -k
            )[-k:]

            for j in idx:

                score = float(scores[j])

                if score > 0:

                    name_best[i].append(
                        (
                            score,
                            chunk.iloc[j]["entity_id"],
                            str(chunk.iloc[j]["country"])
                        )
                    )


        # ----------------------------------------------------
        # ADDRESS
        # ----------------------------------------------------

        chunk_address_matrix = address_vectorizer.transform(
            chunk["address_norm"]
            .fillna("")
            .astype(str)
        )

        address_scores = (
            s1_address_matrix @
            chunk_address_matrix.T
        )

        for i in range(len(s1_eval)):

            scores = address_scores[i].toarray().ravel()

            k = min(
                TOP_K,
                len(scores)
            )

            if k == 0:
                continue

            idx = np.argpartition(
                scores,
                -k
            )[-k:]

            for j in idx:

                score = float(scores[j])

                if score > 0:

                    address_best[i].append(
                        (
                            score,
                            chunk.iloc[j]["entity_id"],
                            str(chunk.iloc[j]["country"])
                        )
                    )


        # ----------------------------------------------------
        # PROGRESS
        # ----------------------------------------------------

        if (
            end % 500_000 == 0
            or end == len(source_df)
        ):

            print(
                f"Processed {end:,}/{len(source_df):,}"
            )


# ------------------------------------------------------------
# 6. CREATE FINAL CANDIDATE SETS
# ------------------------------------------------------------

candidate_sets_eval = []


for i in range(len(s1_eval)):

    s1_country = str(
        s1_eval.iloc[i]["country"]
    )


    # --------------------------------------------------------
    # NAME CANDIDATES
    # --------------------------------------------------------

    name_candidates = [
        x
        for x in name_best[i]
        if x[2] == s1_country
    ]

    name_candidates = sorted(
        name_candidates,
        key=lambda x: x[0],
        reverse=True
    )[:TOP_K]


    # --------------------------------------------------------
    # ADDRESS CANDIDATES
    # --------------------------------------------------------

    address_candidates = [
        x
        for x in address_best[i]
        if x[2] == s1_country
    ]

    address_candidates = sorted(
        address_candidates,
        key=lambda x: x[0],
        reverse=True
    )[:TOP_K]


    # --------------------------------------------------------
    # UNION NAME + ADDRESS
    # --------------------------------------------------------

    candidates = set()

    for item in name_candidates:
        candidates.add(item[1])

    for item in address_candidates:
        candidates.add(item[1])


    candidate_sets_eval.append(
        candidates
    )


# ------------------------------------------------------------
# 7. SUMMARY
# ------------------------------------------------------------

candidate_counts = [
    len(x)
    for x in candidate_sets_eval
]


print("\n")
print("=" * 60)
print("CANDIDATE GENERATION COMPLETED")
print("=" * 60)

print(
    "TOP_K:",
    TOP_K
)

print(
    "Average candidates/S1:",
    round(
        np.mean(candidate_counts),
        2
    )
)

print(
    "Maximum candidates:",
    max(candidate_counts)
)

print(
    "Minimum candidates:",
    min(candidate_counts)
)

print(
    "Total candidate pairs:",
    sum(candidate_counts)
)
```

```python
# ============================================================
# TOP-K = 50 BLOCKING RECALL
# ============================================================

total_true = 0
found_true = 0
missed_true = 0

missed_cases_50 = []

for i, row in s1_eval.reset_index(drop=True).iterrows():

    s1_id = str(row["entity_id"])

    true_ids = ground_truth_eval.get(
        s1_id,
        set()
    )

    candidates = candidate_sets_eval[i]

    found = true_ids & candidates
    missed = true_ids - candidates

    total_true += len(true_ids)
    found_true += len(found)
    missed_true += len(missed)

    for true_id in missed:
        missed_cases_50.append(
            (s1_id, true_id)
        )

recall = (
    found_true / total_true
    if total_true > 0
    else 0
)

print("=" * 60)
print("TOP-K = 50 BLOCKING RECALL")
print("=" * 60)

print("True matches:", total_true)
print("Found:", found_true)
print("Missed:", missed_true)

print(
    "Blocking Recall:",
    round(recall, 4)
)

print(
    "Blocking Recall %:",
    round(recall * 100, 2),
    "%"
)

print(
    "Average candidates/S1:",
    round(
        np.mean(
            [len(x) for x in candidate_sets_eval]
        ),
        2
    )
)

print(
    "Maximum candidates/S1:",
    max(
        len(x)
        for x in candidate_sets_eval
    )
)
```

```python
# ============================================================
# DIAGNOSE THE 16 REMAINING MISSED MATCHES
# ============================================================

from rapidfuzz.fuzz import ratio, token_set_ratio, WRatio
import re

missed_ids = set(
    missed_cases_50_df["true_id"].astype(str)
)

# Find the actual true entities
s2_true = train_s2[
    train_s2["entity_id"].astype(str).isin(missed_ids)
].copy()

s3_true = train_s3[
    train_s3["entity_id"].astype(str).isin(missed_ids)
].copy()

true_entities = pd.concat(
    [s2_true, s3_true],
    ignore_index=True
)

# Merge S1 + true entity
diagnostic_16 = missed_cases_50_df.merge(
    true_entities[
        [
            "entity_id",
            "business_name",
            "business_address",
            "name_norm",
            "address_norm",
            "country"
        ]
    ],
    left_on="true_id",
    right_on="entity_id",
    how="left",
    suffixes=("_s1", "_true")
)


# ------------------------------------------------------------
# Similarity functions
# ------------------------------------------------------------

def get_numbers(text):
    if pd.isna(text):
        return set()
    return set(re.findall(r"\d+", str(text)))


def get_tokens(text):
    if pd.isna(text):
        return set()
    return set(
        str(text).lower().split()
    )


diagnostic_16["name_ratio"] = diagnostic_16.apply(
    lambda r: ratio(
        str(r["s1_name"]),
        str(r["name_norm"])
    ),
    axis=1
)

diagnostic_16["name_token_ratio"] = diagnostic_16.apply(
    lambda r: token_set_ratio(
        str(r["s1_name"]),
        str(r["name_norm"])
    ),
    axis=1
)

diagnostic_16["name_WRatio"] = diagnostic_16.apply(
    lambda r: WRatio(
        str(r["s1_name"]),
        str(r["name_norm"])
    ),
    axis=1
)

diagnostic_16["address_ratio"] = diagnostic_16.apply(
    lambda r: ratio(
        str(r["s1_address"]),
        str(r["address_norm"])
    ),
    axis=1
)

diagnostic_16["shared_numbers"] = diagnostic_16.apply(
    lambda r:
        get_numbers(r["s1_address"]) &
        get_numbers(r["address_norm"]),
    axis=1
)

diagnostic_16["shared_address_tokens"] = diagnostic_16.apply(
    lambda r:
        get_tokens(r["s1_address"]) &
        get_tokens(r["address_norm"]),
    axis=1
)


# ------------------------------------------------------------
# Display
# ------------------------------------------------------------

display(
    diagnostic_16[
        [
            "s1_id",
            "true_id",
            "s1_name",
            "business_name",
            "s1_address",
            "business_address",
            "name_ratio",
            "name_token_ratio",
            "name_WRatio",
            "address_ratio",
            "shared_numbers",
            "shared_address_tokens"
        ]
    ]
)
```

```python
# ============================================================
# TEST TF-IDF SIMILARITY FOR THE 16 MISSED TRUE PAIRS
# ============================================================

# Prepare true names and addresses
true_names = (
    diagnostic_16["name_norm"]
    .fillna("")
    .astype(str)
    .tolist()
)

true_addresses = (
    diagnostic_16["address_norm"]
    .fillna("")
    .astype(str)
    .tolist()
)

# Transform the 16 true records using the SAME vectorizers
true_name_matrix = name_vectorizer.transform(true_names)
true_address_matrix = address_vectorizer.transform(true_addresses)

# Calculate S1 <-> true-pair similarity
name_tfidf_scores = []

address_tfidf_scores = []

for i in range(len(diagnostic_16)):

    name_score = (
        s1_name_matrix[i] @
        true_name_matrix[i].T
    ).toarray()[0][0]

    address_score = (
        s1_address_matrix[i] @
        true_address_matrix[i].T
    ).toarray()[0][0]

    name_tfidf_scores.append(float(name_score))
    address_tfidf_scores.append(float(address_score))


diagnostic_16["name_tfidf"] = name_tfidf_scores
diagnostic_16["address_tfidf"] = address_tfidf_scores

diagnostic_16["combined_tfidf"] = (
    0.5 * diagnostic_16["name_tfidf"] +
    0.5 * diagnostic_16["address_tfidf"]
)


# ------------------------------------------------------------
# DISPLAY
# ------------------------------------------------------------

display(
    diagnostic_16[
        [
            "s1_id",
            "true_id",
            "s1_name",
            "business_name",
            "name_ratio",
            "name_tfidf",
            "address_ratio",
            "address_tfidf",
            "combined_tfidf"
        ]
    ]
)
```

```python
# ============================================================
# NORMALIZE TEXT
# ============================================================

def normalize_text_unicode(x):

    if pd.isna(x):
        return ""

    x = str(x).lower()

    replacements = {
        "private limited": "pvt ltd",
        "private ltd": "pvt ltd",
        "corporation": "corp",
        "incorporated": "inc",
        "limited": "ltd",
        "company": "co",
        "street": "st",
        "road": "rd",
        "avenue": "ave",
        "boulevard": "blvd",
        "drive": "dr",
        "lane": "ln",
        "highway": "hwy",
        "parkway": "pkwy",
        "apartment": "apt",
        "suite": "ste",
        "place": "pl",
        "court": "ct"
    }

    for old, new in replacements.items():
        x = x.replace(old, new)

    x = re.sub(
        r"[^\w\s]",
        " ",
        x,
        flags=re.UNICODE
    )

    x = re.sub(
        r"\s+",
        " ",
        x
    ).strip()

    return x


test_s1["name_norm"] = (
    test_s1["business_name"]
    .fillna("")
    .apply(normalize_text_unicode)
)

test_s1["address_norm"] = (
    test_s1["business_address"]
    .fillna("")
    .apply(normalize_text_unicode)
)

print("Normalization completed.")
print(test_s1.shape)
```

```python
import pandas as pd
import numpy as np
import time
from rapidfuzz import fuzz
from pathlib import Path

CANDIDATE_FILE = Path("output/candidate_pairs.tsv")

THRESHOLD = 0.10
CHUNK_SIZE = 100_000

start = time.time()
total_candidates = 0
total_matches = 0

# Create temporary scored-pairs table
conn.execute("DROP TABLE IF EXISTS scored_pairs")

conn.execute("""
CREATE TABLE scored_pairs (
    source1_entity_id TEXT,
    candidate_entity_id TEXT
)
""")

conn.commit()

print("Starting candidate scoring...")

for chunk_no, chunk in enumerate(
    pd.read_csv(
        CANDIDATE_FILE,
        sep="\t",
        dtype=str,
        chunksize=CHUNK_SIZE
    ),
    start=1
):

    # Join candidate pairs with S1 and candidate entity information
    ids = chunk["candidate_entity_id"].tolist()

    # Get candidate entity information in batches
    placeholders = ",".join(["?"] * len(ids))

    candidate_data = pd.read_sql_query(
        f"""
        SELECT
            entity_id,
            name_norm,
            address_norm,
            country
        FROM entities
        WHERE entity_id IN ({placeholders})
        """,
        conn,
        params=ids
    )

    candidate_data = candidate_data.drop_duplicates("entity_id")

    chunk = chunk.merge(
        candidate_data,
        left_on="candidate_entity_id",
        right_on="entity_id",
        how="inner"
    )

    # Get S1 information
    s1_data = pd.read_sql_query(
        f"""
        SELECT
            entity_id,
            name_norm,
            address_norm,
            country
        FROM test_s1_index
        WHERE entity_id IN ({",".join(["?"] * chunk["source1_entity_id"].nunique())})
        """,
        conn,
        params=chunk["source1_entity_id"].unique().tolist()
    )

    s1_data = s1_data.drop_duplicates("entity_id")

    chunk = chunk.merge(
        s1_data,
        left_on="source1_entity_id",
        right_on="entity_id",
        how="inner",
        suffixes=("_candidate", "_s1")
    )

    # Model features
    chunk["name_exact"] = (
        chunk["name_norm_s1"] == chunk["name_norm_candidate"]
    ).astype(int)

    chunk["address_exact"] = (
        chunk["address_norm_s1"] == chunk["address_norm_candidate"]
    ).astype(int)

    chunk["name_similarity"] = [
        fuzz.ratio(a, b) / 100
        for a, b in zip(
            chunk["name_norm_s1"],
            chunk["name_norm_candidate"]
        )
    ]

    chunk["address_similarity"] = [
        fuzz.ratio(a, b) / 100
        for a, b in zip(
            chunk["address_norm_s1"],
            chunk["address_norm_candidate"]
        )
    ]

    chunk["country_match"] = (
        chunk["country_s1"].str.lower() ==
        chunk["country_candidate"].str.lower()
    ).astype(int)

    features = [
        "name_exact",
        "name_similarity",
        "address_exact",
        "address_similarity",
        "country_match"
    ]

    probabilities = model.predict_proba(
        chunk[features]
    )[:, 1]

    chunk["probability"] = probabilities

    matches = chunk[
        chunk["probability"] >= THRESHOLD
    ][[
        "source1_entity_id",
        "candidate_entity_id"
    ]]

    if len(matches):
        matches.to_sql(
            "scored_pairs",
            conn,
            if_exists="append",
            index=False
        )

        total_matches += len(matches)

    total_candidates += len(chunk)

    if chunk_no % 10 == 0:
        elapsed = time.time() - start
        print(
            f"Processed: {total_candidates:,} | "
            f"Matches: {total_matches:,} | "
            f"Time: {elapsed/60:.1f} min"
        )

conn.commit()

elapsed = time.time() - start

print("\n================================")
print("SCORING COMPLETE")
print("================================")
print("Candidates scored:", f"{total_candidates:,}")
print("Predicted matches:", f"{total_matches:,}")
print("Time:", round(elapsed / 60, 2), "minutes")
```

```python
import pandas as pd
import numpy as np
import time
from rapidfuzz import fuzz
from pathlib import Path

CANDIDATE_FILE = Path("output/candidate_pairs.tsv")

THRESHOLD = 0.10
CHUNK_SIZE = 100_000

start = time.time()
total_candidates = 0
total_matches = 0

# Reset temporary tables
conn.execute("DROP TABLE IF EXISTS candidate_chunk")
conn.execute("DROP TABLE IF EXISTS scored_pairs")

conn.execute("""
CREATE TEMP TABLE candidate_chunk (
    source1_entity_id TEXT,
    candidate_entity_id TEXT,
    PRIMARY KEY (source1_entity_id, candidate_entity_id)
)
""")

conn.execute("""
CREATE TABLE scored_pairs (
    source1_entity_id TEXT,
    candidate_entity_id TEXT
)
""")

conn.commit()

print("Starting candidate scoring...")

for chunk_no, chunk in enumerate(
    pd.read_csv(
        CANDIDATE_FILE,
        sep="\t",
        dtype=str,
        chunksize=CHUNK_SIZE
    ),
    start=1
):

    # Clear previous chunk
    conn.execute("DELETE FROM candidate_chunk")

    # Insert candidate pairs into SQLite
    conn.executemany(
        """
        INSERT OR IGNORE INTO candidate_chunk
        (source1_entity_id, candidate_entity_id)
        VALUES (?, ?)
        """,
        list(
            chunk[
                ["source1_entity_id", "candidate_entity_id"]
            ].itertuples(index=False, name=None)
        )
    )

    conn.commit()

    # Retrieve both S1 and candidate records through JOIN
    data = pd.read_sql_query(
        """
        SELECT
            cp.source1_entity_id,
            cp.candidate_entity_id,

            s1.name_norm AS s1_name,
            s1.address_norm AS s1_address,
            s1.country AS s1_country,

            e.name_norm AS candidate_name,
            e.address_norm AS candidate_address,
            e.country AS candidate_country

        FROM candidate_chunk cp

        JOIN test_s1_index s1
          ON cp.source1_entity_id = s1.entity_id

        JOIN entities e
          ON cp.candidate_entity_id = e.entity_id
        """,
        conn
    )

    if len(data) == 0:
        continue

    # -----------------------------
    # Model features
    # -----------------------------

    data["name_exact"] = (
        data["s1_name"] == data["candidate_name"]
    ).astype(int)

    data["address_exact"] = (
        data["s1_address"] == data["candidate_address"]
    ).astype(int)

    data["country_match"] = (
        data["s1_country"].str.lower()
        ==
        data["candidate_country"].str.lower()
    ).astype(int)

    # RapidFuzz similarities
    data["name_similarity"] = [
        fuzz.ratio(a, b) / 100
        for a, b in zip(
            data["s1_name"],
            data["candidate_name"]
        )
    ]

    data["address_similarity"] = [
        fuzz.ratio(a, b) / 100
        for a, b in zip(
            data["s1_address"],
            data["candidate_address"]
        )
    ]

    features = [
        "name_exact",
        "name_similarity",
        "address_exact",
        "address_similarity",
        "country_match"
    ]

    # Model prediction
    probabilities = model.predict_proba(
        data[features]
    )[:, 1]

    data["probability"] = probabilities

    matches = data[
        data["probability"] >= THRESHOLD
    ][[
        "source1_entity_id",
        "candidate_entity_id"
    ]]

    if len(matches):
        conn.executemany(
            """
            INSERT INTO scored_pairs
            (source1_entity_id, candidate_entity_id)
            VALUES (?, ?)
            """,
            list(
                matches.itertuples(
                    index=False,
                    name=None
                )
            )
        )

        conn.commit()

        total_matches += len(matches)

    total_candidates += len(data)

    if chunk_no % 10 == 0:
        elapsed = time.time() - start
        print(
            f"Chunk {chunk_no:,} | "
            f"Candidates: {total_candidates:,} | "
            f"Matches: {total_matches:,} | "
            f"Time: {elapsed/60:.1f} min"
        )

conn.commit()

elapsed = time.time() - start

print("\n================================")
print("SCORING COMPLETE")
print("================================")
print("Candidates scored:", f"{total_candidates:,}")
print("Predicted matches:", f"{total_matches:,}")
print("Time:", round(elapsed / 60, 2), "minutes")
print("================================")
```

```python
import sqlite3
import pandas as pd
from rapidfuzz import fuzz
import time

DB = "test_index/entity_index.db"
CANDIDATES = "output/candidate_pairs.tsv"
TEST_S1 = "/Users/apple/Downloads/student_resource 2/dataset/test/test_source1.tsv"
OUT = "output/matching_results_v2.tsv"

t0 = time.time()

# -----------------------------
# Load Test Source 1
# -----------------------------
s1 = pd.read_csv(
    TEST_S1,
    sep="\t",
    dtype=str,
    usecols=["entity_id", "business_name", "business_address"]
).fillna("")

s1 = s1.rename(columns={"entity_id": "source1_entity_id"})

s1["name_norm"] = s1["business_name"].str.lower()
s1["address_norm"] = s1["business_address"].str.lower()

s1_lookup = s1.set_index("source1_entity_id")[
    ["name_norm", "address_norm"]
]

# -----------------------------
# SQLite connection
# -----------------------------
conn = sqlite3.connect(DB)

results = []
total_pairs = 0
total_matches = 0

# -----------------------------
# Process candidate file
# -----------------------------
for chunk_no, chunk in enumerate(
    pd.read_csv(
        CANDIDATES,
        sep="\t",
        dtype=str,
        chunksize=5000
    ),
    1
):

    # Convert grouped candidates into individual pairs
    pairs = chunk.assign(
        candidate_entity_id=
        chunk["candidate_entity_ids"]
        .fillna("")
        .str.split(",")
    ).explode("candidate_entity_id")

    pairs = pairs[
        pairs["candidate_entity_id"].notna()
        & (pairs["candidate_entity_id"] != "")
    ].copy()

    if pairs.empty:
        continue

    # Add S1 normalized text
    pairs = pairs.join(
        s1_lookup,
        on="source1_entity_id"
    )

    # -----------------------------------------
    # Put candidate IDs into temporary SQLite table
    # -----------------------------------------
    conn.execute("DROP TABLE IF EXISTS candidate_chunk")

    unique_ids = pairs[["candidate_entity_id"]].drop_duplicates()

    unique_ids.to_sql(
        "candidate_chunk",
        conn,
        index=False,
        if_exists="replace"
    )

    # -----------------------------------------
    # JOIN instead of huge IN(...)
    # -----------------------------------------
    cand_info = pd.read_sql_query(
        """
        SELECT
            c.candidate_entity_id,
            e.name_norm AS candidate_name_norm,
            e.address_norm AS candidate_address_norm
        FROM candidate_chunk c
        INNER JOIN entities e
            ON e.entity_id = c.candidate_entity_id
        """,
        conn
    )

    pairs = pairs.merge(
        cand_info,
        on="candidate_entity_id",
        how="inner"
    )

    # -----------------------------------------
    # Similarities
    # -----------------------------------------
    pairs["name_similarity"] = [
        fuzz.ratio(a, b) / 100.0
        for a, b in zip(
            pairs["name_norm"],
            pairs["candidate_name_norm"]
        )
    ]

    pairs["address_similarity"] = [
        fuzz.ratio(a, b) / 100.0
        for a, b in zip(
            pairs["address_norm"],
            pairs["candidate_address_norm"]
        )
    ]

    pairs["address_exact"] = (
        (pairs["address_norm"] != "") &
        (pairs["address_norm"] == pairs["candidate_address_norm"])
    )

    # -----------------------------------------
    # BEST VALIDATED RULE
    # -----------------------------------------
    keep = (
        pairs["address_exact"]
        |
        (
            (pairs["name_similarity"] >= 0.80) &
            (pairs["address_similarity"] >= 0.60)
        )
    )

    selected = pairs.loc[
        keep,
        ["source1_entity_id", "candidate_entity_id"]
    ]

    if not selected.empty:
        results.append(selected)

    total_pairs += len(pairs)
    total_matches += len(selected)

    if chunk_no % 20 == 0:
        print(
            f"Chunk {chunk_no} | "
            f"Pairs: {total_pairs:,} | "
            f"Matches: {total_matches:,} | "
            f"Time: {(time.time()-t0)/60:.1f} min"
        )

# -----------------------------
# Combine results
# -----------------------------
if results:
    pred = pd.concat(results, ignore_index=True)

    pred = (
        pred.groupby("source1_entity_id")["candidate_entity_id"]
        .agg(",".join)
        .reset_index()
    )

    pred = pred.rename(
        columns={"candidate_entity_id": "matched_entity_ids"}
    )
else:
    pred = pd.DataFrame(
        columns=["source1_entity_id", "matched_entity_ids"]
    )

# -----------------------------
# Ensure every S1 is present
# -----------------------------
final = s1[["source1_entity_id"]].merge(
    pred,
    on="source1_entity_id",
    how="left"
)

final["matched_entity_ids"] = (
    final["matched_entity_ids"].fillna("")
)

# -----------------------------
# Save V2
# -----------------------------
final.to_csv(
    OUT,
    sep="\t",
    index=False
)

print("\n==============================")
print("DONE")
print("==============================")
print("Rows:", len(final))
print("Expected:", len(s1))
print(
    "Non-empty:",
    (final["matched_entity_ids"] != "").sum()
)
print(
    "Empty:",
    (final["matched_entity_ids"] == "").sum()
)
print("Saved:", OUT)
print(
    "Time:",
    round((time.time()-t0)/60, 1),
    "minutes"
)
```

```python
import pandas as pd
import time

t0 = time.time()

MATCH = "output/matching_results.tsv"
CAND = "output/candidate_pairs.tsv"
OUT = "output/candidate_pairs_compact.tsv"

# Load final matches
matches = pd.read_csv(
    MATCH,
    sep="\t",
    dtype=str,
    keep_default_na=False
)

# Load candidate file
candidates = pd.read_csv(
    CAND,
    sep="\t",
    dtype=str,
    keep_default_na=False
)

# Create lookup of final matched IDs
match_map = dict(
    zip(
        matches["source1_entity_id"],
        matches["matched_entity_ids"]
    )
)

# Keep ONLY candidates that are actually in our final matching result
def filter_candidates(row):

    matched = match_map.get(
        row["source1_entity_id"],
        ""
    )

    if not matched:
        return ""

    matched_set = set(matched.split(","))

    ids = [
        x for x in row["candidate_entity_ids"].split(",")
        if x and x in matched_set
    ]

    return ",".join(ids)

candidates["candidate_entity_ids"] = candidates.apply(
    filter_candidates,
    axis=1
)

# Save
candidates.to_csv(
    OUT,
    sep="\t",
    index=False
)

print("DONE")
print("Rows:", len(candidates))
print(
    "Non-empty:",
    (candidates["candidate_entity_ids"] != "").sum()
)
print(
    "Empty:",
    (candidates["candidate_entity_ids"] == "").sum()
)
print(
    "Original candidate rows:",
    len(candidates)
)
print("Saved:", OUT)
print("Time:", round((time.time()-t0)/60, 2), "min")
```
